# **Phần 2: Phân tích câu và Trực quan hóa**
## **2.1. Tải mô hình và phân tích câu**

```python
import spacy
from spacy import displacy
import en_core_web_md

# Tải mô hình tiếng Anh đã cài đặt
# Sử dụng en_core_web_md vì nó chứa các vector từ và cây cú pháp đầy đủ
nlp = en_core_web_md.load()
# Câu ví dụ
text = "The quick brown fox jumps over the lazy dog."
# Phân tích câu với pipeline của spaCy
doc = nlp(text)
```

## **2.2. Trực quan hóa cây phụ thuộc**

```python
displacy.serve(doc, style="dep")
```

• Từ nào là gốc (ROOT) của câu : jumps

• jumps có những từ phụ thuộc (dependent) và quan hệ: fox(usubj) , over(prep)

• fox là head của những từ: The, quick, brown

```python
# Lấy một câu khác để phân tích
text = "Apple is looking at buying U.K. startup for $1 billion"
doc = nlp(text)
# In ra thông tin của từng token
print(f"{'TEXT':<12} | {'DEP':<10} | {'HEAD TEXT':<12} | {'HEAD POS':<8} | {'CHILDREN'}")
print("-" * 70)

for token in doc:
# Trích xuất các thuộc tính
  children = [child.text for child in token.children]
  print(f"{token.text:<12} | {token.dep_:<10} | {token.head.text:<12} | {token.head.pos_:<8}")
```

# **Phần 4: Duyệt cây phụ thuộc để trích xuất thông tin**
## **4.1. Bài toán: Tìm chủ ngữ và tân ngữ của một động từ**

```python
text = "The cat chased the mouse and the dog watched them."
doc = nlp(text)
for token in doc:
# Chỉ tìm các động từ
  if token.pos_ == "VERB":
    verb = token.text
    subject = ""

    obj = ""
    # Tìm chủ ngữ (nsubj) và tân ngữ (dobj) trong các con của động từ
    for child in token.children:
      if child.dep_ == "nsubj":
        subject = child.text
      if child.dep_ == "dobj":
        obj = child.text
    if subject and obj:
      print(f"Found Triplet: ({subject}, {verb}, {obj})")
```

# **4.2. Bài toán: Tìm các tính từ bổ nghĩa cho một danh từ**

```python
text = "The big, fluffy white cat is sleeping on the warm mat."
doc = nlp(text)
for token in doc:
# Chỉ tìm các danh từ
  if token.pos_ == "NOUN":
    adjectives = []
    # Tìm các tính từ bổ nghĩa (amod) trong các con của danh từ
    for child in token.children:
      if child.dep_ == "amod":
        adjectives.append(child.text)
    if adjectives:
      print(f"Danh từ '{token.text}' được bổ nghĩa bởi các tính từ: {adjectives}")
```

# **Phần 5: Bài tập tự luyện**
## **Bài 1: Tìm động từ chính của câu**

```python
def find_main_verb(doc):
  for token in doc:
    if token.dep_ == "ROOT":
      return token
  return None

doc_ex1 = nlp("The quick brown fox jumps over the lazy dog.")
main_verb = find_main_verb(doc_ex1)
print(f"Động từ chính: {main_verb.text} (POS: {main_verb.pos_})")
```

## **Bài 2: Trích xuất các cụm danh từ (Noun Chunks)**

```python
def get_subtree_text(token):
    """Hàm đệ quy để lấy toàn bộ từ ngữ trong nhánh của token theo đúng thứ tự"""
    # Lấy token hiện tại và tất cả con cháu của nó (subtree)
    # Sắp xếp lại theo chỉ số (index) xuất hiện trong câu để cụm từ có nghĩa
    words = sorted([t for t in token.subtree], key=lambda t: t.i)
    return " ".join([t.text for t in words])

def custom_noun_chunks(doc):
    chunks = []
    for token in doc:
        # Tìm các danh từ (NOUN) hoặc danh từ riêng (PROPN)
        # Và đảm bảo nó không phải là con của một danh từ khác (để tránh lấy cụm chồng cụm)
        if token.pos_ in ["NOUN", "PROPN"]:
            # Logic đơn giản: Lấy toàn bộ nhánh con của danh từ này
            # Lưu ý: Đây là cách đơn giản, thực tế spaCy có logic phức tạp hơn
            chunk_text = get_subtree_text(token)
            chunks.append(chunk_text)
    return chunks

# Test
doc_ex2 = nlp("The quick brown fox jumps over the lazy dog")
print(f"Bài 2 - Các cụm danh từ (tự viết): {custom_noun_chunks(doc_ex2)}")
# So sánh với spaCy có sẵn
print(f"Bài 2 - Các cụm danh từ (spaCy): {[chunk.text for chunk in doc_ex2.noun_chunks]}")
```

## **Bài 3: Tìm đường đi ngắn nhất trong cây**

```python
def get_path_to_root(token):
    """Trả về danh sách token từ token hiện tại lên đến ROOT."""
    path = [token]
    current = token
    # Duyệt ngược lên head cho đến khi gặp chính nó (Root head là chính nó)
    # hoặc check current.dep_ == 'ROOT'
    while current.head != current:
        current = current.head
        path.append(current)
    return path

# Test
doc_ex3 = nlp("The quick brown fox jumps over the lazy dog")
# Giả sử ta muốn tìm đường đi từ từ "startup" lên gốc
token_startup = doc_ex3[2] # vị trí của từ startup
path = get_path_to_root(token_startup)

print("Bài 3 - Đường đi từ 'startup' đến ROOT:")
# In ra đường đi với mũi tên
path_text = " -> ".join([t.text for t in path])
print(path_text)
```
